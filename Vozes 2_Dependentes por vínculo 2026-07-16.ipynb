## Query adaptada de consulta no DW-SIAPE

Consulta construída [neste relatório do DW-SIAPE](http://dw.siapenet.gov.br:80/dwsiape/servlet/mstrWeb?evt=4058&src=mstrWeb.4058&_subscriptionID=81D721D07C4FED988E5175BFAF8FF111&reportViewMode=1&Server=161.148.236.209&Project=PESSOAL%20-%20Vers%C3%A3o%203.1_b&Port=0&share=1). 

Na página do reltório, a query foi obtida em Ferramentas > Página Detalhes do Relatório e adaptada para as referências à base `mgi-ouro`.bd_dwsiape. 

O relatório foi construído com filtro em servidores que se auto-declaram indígenas, para diminuir o volume da consulta. Na query adaptada abaixo, o filtro para indígenas foi removido.

```r
%sql
create table `mgi-ouro`.`bd_coest`.`vozes2_dw_dependentes` as
select	a11.MES,
	a110.NOME_MES,
    a11.VINCULO_SERV,
	a12.ORGAO_VINC AS COD_ORGAO,
	a112.NOME_ORGAO_VINC AS SG_ORGAO,
	a112.NOME_ORGAO_VINC_COMPLETO NO_ORGAO,
	a18.GRUPO_SIT_VINC_SERV,
	a113.NOME_GRUPO_SIT_VINC_SERV,
	a13.SERVIDOR_PUBLICO,
	a14.NOME_SERVIDOR_PUBLICO,
	a14.CPF,
	LPAD(CAST(a14.CPF AS STRING), 11, '0') as CustCol_15,
	a14.NOME2_SERVIDOR_PUBLICO,
	a16.MAE_DEPENDENTE,
	a16.NOME_DEPENDENTE,
	a16.CPF_DEPENDENTE,
	a15.PARENTESCO_DEPENDENTE,
	a111.NOME_PARENTESCO_DEPENDENTE,
	a15.CONDICAO_DEPENDENTE,
	a19.NOME_CONDICAO_DEPENDENTE,
	a16.DT_NASC_DEPENDENTE,
	sum(a11.QTDE_VINCULO_SER_PES) as WJXBFS1
from	`mgi-ouro`.`bd_dwsiape`.`WF_PESSOAL301_TAB`	a11
	join	`mgi-ouro`.`bd_dwsiape`.`UORG_VINC`	a12
	  on 	(a11.UORG_VINC = a12.UORG_VINC)
	join	`mgi-ouro`.`bd_dwsiape`.`VINCULO_SERV`	a13
	  on 	(a11.VINCULO_SERV = a13.VINCULO_SERV)
	join	`mgi-ouro`.`bd_dwsiape`.`SERVIDOR_PUBLICO`	a14
	  on 	(a13.SERVIDOR_PUBLICO = a14.SERVIDOR_PUBLICO)
	join	`mgi-ouro`.`bd_dwsiape`.`VINCULO_BENEF_DEPEND`	a15
	  on 	(a11.VINCULO_SERV = a15.VINCULO_SERV)
	join	`mgi-ouro`.`bd_dwsiape`.`DEPENDENTE`	a16
	  on 	(a15.DEPENDENTE = a16.DEPENDENTE)
	join	`mgi-ouro`.`bd_dwsiape`.`ESTADO_VINC_SERV`	a17
	  on 	(a11.ESTADO_VINC_SERV = a17.ESTADO_VINC_SERV)
	join	`mgi-ouro`.`bd_dwsiape`.`SITUACAO_FUNC_VINC_SERV`	a18
	  on 	(a11.SITUACAO_FUNC_VINC_SERV = a18.SITUACAO_FUNC_VINC_SERV)
	join	`mgi-ouro`.`bd_dwsiape`.`CONDICAO_DEPENDENTE`	a19
	  on 	(a15.CONDICAO_DEPENDENTE = a19.CONDICAO_DEPENDENTE)
	join	`mgi-ouro`.`bd_dwsiape`.`MES`	a110
	  on 	(a11.MES = a110.MES)
	join	`mgi-ouro`.`bd_dwsiape`.`PARENTESCO_DEPENDENTE`	a111
	  on 	(a15.PARENTESCO_DEPENDENTE = a111.PARENTESCO_DEPENDENTE)
	join	`mgi-ouro`.`bd_dwsiape`.`ORGAO_VINC`	a112
	  on 	(a12.ORGAO_VINC = a112.ORGAO_VINC)
	join	`mgi-ouro`.`bd_dwsiape`.`GRUPO_SIT_VINC_SERV`	a113
	  on 	(a18.GRUPO_SIT_VINC_SERV = a113.GRUPO_SIT_VINC_SERV)
where	((a17.EXCLUSAO_VINC_SERV in (0)
 or a18.GRUPO_SIT_VINC_SERV in (3))
 and a11.MES in (202510)
 and a12.GRUPO_NAT_JURIDICA_CNNJ not in (2)
 and a12.NATUREZA_JURIDICA_CNNJ not in (102)
 and a12.ORGAO_VINC not in (99072)
 and a18.REGIME_JURIDICO not in ('RMI', 'ETE', 'ETG')
 and a11.SITUACAO_FUNC_VINC_SERV not in (26, 27, 57, 58, 53, 52)
 ---and a14.COR_ORIGEM_ETNICA in (5)
 and a18.GRUPO_SIT_VINC_SERV in (1))
group by	a11.MES,
	a11.VINCULO_SERV,
    a110.NOME_MES,
	a12.ORGAO_VINC,
	a112.NOME_ORGAO_VINC,
	a112.NOME_ORGAO_VINC_COMPLETO,
	a18.GRUPO_SIT_VINC_SERV,
	a113.NOME_GRUPO_SIT_VINC_SERV,
	a13.SERVIDOR_PUBLICO,
	a14.NOME_SERVIDOR_PUBLICO,
	a14.CPF,
	LPAD(CAST(a14.CPF AS STRING), 11, '0'),
	a14.NOME2_SERVIDOR_PUBLICO,
	a16.MAE_DEPENDENTE,
	a16.NOME_DEPENDENTE,
	a16.CPF_DEPENDENTE,
	a15.PARENTESCO_DEPENDENTE,
	a111.NOME_PARENTESCO_DEPENDENTE,
	a15.CONDICAO_DEPENDENTE,
	a19.NOME_CONDICAO_DEPENDENTE,
	a16.DT_NASC_DEPENDENTE
```

## Obtendo idade dos dependentes

### Calculando idade do dependente em 30/10/2025

Além do cálculo da idade, foi obtida uma flag que assume 1 se o dependente é menor de 18 anos, 0 caso contrário

```r
%python
from pyspark.sql.functions import col, concat, to_date, last_day, datediff, round, lit, substring

# MES é double (ex: 202510.0) → cast para long antes de converter para string
# DT_NASC_DEPENDENTE é timestamp string no formato 'yyyy-MM-dd HH:mm:ss.S' → pega só os 10 primeiros caracteres
df_idade = df_base.withColumn(
    "MES_DATA",
    last_day(to_date(concat(col("MES").cast("long").cast("string"), lit("01")), "yyyyMMdd"))
).withColumn(
    "DT_NASC_DEPENDENTE_DATA",
    to_date(substring(col("DT_NASC_DEPENDENTE"), 1, 10), "yyyy-MM-dd")
).withColumn(
    "idade_dependente_dias",
    datediff(col("MES_DATA"), col("DT_NASC_DEPENDENTE_DATA"))
).withColumn(
    "idade_dependente_anos",
    round(col("idade_dependente_dias") / lit(365.25), 2)
)


from pyspark.sql.functions import when

df_idade = df_idade.withColumn(
    "dependente_menor_18",
    when(col("idade_dependente_anos") < 18, 1).otherwise(0)
)
display(df_idade)
```

### Checagem 1: prevalências de graus de parentesco dos dependentes menores X dependentes maiores

```r
%python
from pyspark.sql.functions import count, countDistinct, sum as spark_sum, col, round

# Agrega por flag de maioridade, parentesco e nome do parentesco
df_agregado = df_idade.groupBy(
    "dependente_menor_18",
    "PARENTESCO_DEPENDENTE",
    "NOME_PARENTESCO_DEPENDENTE"
).agg(
    count("*").alias("n_linhas"),
    countDistinct("CPF_DEPENDENTE").alias("n_dependentes")
)

# Calcula o total de linhas por flag de maioridade
df_totais = df_agregado.groupBy("dependente_menor_18").agg(
    spark_sum("n_linhas").alias("total_flag")
)

# Junta para calcular o percentual
df_final = df_agregado.join(
    df_totais,
    on="dependente_menor_18",
    how="left"
).withColumn(
    "percentual_linhas_flag",
    round(col("n_linhas") / col("total_flag") * 100, 2)
)

df_final = df_final.orderBy(
    col("dependente_menor_18").asc(),
    col("percentual_linhas_flag").desc()
)

p_menores_filhos = df_final.filter(col("dependente_menor_18") == 1).select("percentual_linhas_flag").filter(col("PARENTESCO_DEPENDENTE") == 8).collect()[0][0]

display(df_final)
```

```r
%python
displayHTML(f"<p>- Dentre os registros de dependentes MENORES de 18 anos, as maiores ocorrências de parentesco são:</p>\n\n")


top3_menores = df_final.filter(col("dependente_menor_18") == 1).select("percentual_linhas_flag", "NOME_PARENTESCO_DEPENDENTE").limit(3).collect()

html_top3 = "<ul>"
for row in top3_menores:
    html_top3 += f"<li>{row['NOME_PARENTESCO_DEPENDENTE']}: <b>{row['percentual_linhas_flag']}%</b></li>"
html_top3 += "</ul>"

displayHTML(html_top3)


displayHTML(f"<p>- Dentre os registros de dependentes MAIORES de 18 anos, as maiores ocorrências de parentesco são:</p>\n\n")


top3_maiores = df_final.filter(col("dependente_menor_18") == 0).select("percentual_linhas_flag", "NOME_PARENTESCO_DEPENDENTE").limit(3).collect()

html_top3 = "<ul>"
for row in top3_maiores:
    html_top3 += f"<li>{row['NOME_PARENTESCO_DEPENDENTE']}: <b>{row['percentual_linhas_flag']}%</b></li>"
html_top3 += "</ul>"

displayHTML(html_top3)
```

Checagem 2: prevalências de condições dos dependentes menores X dependentes maiores

```r
%python
from pyspark.sql.functions import count, countDistinct, sum as spark_sum, col, round

# Agrega por flag de maioridade, parentesco e nome do parentesco
df_agregado = df_idade.groupBy(
    "dependente_menor_18",
    "CONDICAO_DEPENDENTE",
    "NOME_CONDICAO_DEPENDENTE"
).agg(
    count("*").alias("n_linhas"),
    countDistinct("CPF_DEPENDENTE").alias("n_dependentes")
)

# Calcula o total de linhas por flag de maioridade
df_totais = df_agregado.groupBy("dependente_menor_18").agg(
    spark_sum("n_linhas").alias("total_flag")
)

# Junta para calcular o percentual
df_final = df_agregado.join(
    df_totais,
    on="dependente_menor_18",
    how="left"
).withColumn(
    "percentual_linhas_flag",
    round(col("n_linhas") / col("total_flag") * 100, 2)
)

df_final = df_final.orderBy(
    col("dependente_menor_18").asc(),
    col("percentual_linhas_flag").desc()
)

p_sem_exigencia = df_final.filter(col("dependente_menor_18") == 1).select("percentual_linhas_flag").filter(col("CONDICAO_DEPENDENTE") == 32).collect()[0][0]

p_3mais_maior = df_final.filter(col("dependente_menor_18") == 0).select("percentual_linhas_flag").collect()[0][0:3]

display(df_final)

print(p_sem_exigencia)
print(p_3mais_maior)
```

```r
%python
displayHTML(f"<p>- Dentre os registros de dependentes MENORES de 18 anos, as maiores ocorrências de condição são:</p>\n\n")


top3_menores = df_final.filter(col("dependente_menor_18") == 1).select("percentual_linhas_flag", "NOME_CONDICAO_DEPENDENTE").limit(3).collect()

html_top3 = "<ul>"
for row in top3_menores:
    html_top3 += f"<li>{row['NOME_CONDICAO_DEPENDENTE']}: <b>{row['percentual_linhas_flag']}%</b></li>"
html_top3 += "</ul>"

displayHTML(html_top3)


displayHTML(f"<p>- Dentre os registros de dependentes MAIORES de 18 anos, as maiores ocorrências de condição são:</p>\n\n")


top3_maiores = df_final.filter(col("dependente_menor_18") == 0).select("percentual_linhas_flag", "NOME_CONDICAO_DEPENDENTE").limit(3).collect()

html_top3 = "<ul>"
for row in top3_maiores:
    html_top3 += f"<li>{row['NOME_CONDICAO_DEPENDENTE']}: <b>{row['percentual_linhas_flag']}%</b></li>"
html_top3 += "</ul>"

displayHTML(html_top3)
```

## Agregando por 'vínculo' (par CPF x órgão)

```r
%python
# Agrega número de vínculos por órgão
df_vinculo = df_idade.groupBy(
    "COD_ORGAO",
    "SG_ORGAO",
    "NO_ORGAO",
    "CPF"
).agg(
    countDistinct("VINCULO_SERV").alias("nvinculos_cpf_orgao"),
    countDistinct("CPF").alias("nrep_cpf")
).orderBy(
    col("SG_ORGAO")
)

df_dpen_vinculo = df_idade.groupBy(
    "COD_ORGAO",
    "SG_ORGAO",
    "NO_ORGAO",
    "CPF",
    "dependente_menor_18",
).agg(
    countDistinct("CPF_DEPENDENTE").alias("n_dependentes")
).orderBy(
    col("SG_ORGAO"),
    col("dependente_menor_18").asc()
)

df_vinculo = df_vinculo.join(
    df_dpen_vinculo,
    on=["COD_ORGAO", "SG_ORGAO", "NO_ORGAO","CPF"],
    how="left"
).groupBy("COD_ORGAO",
    "SG_ORGAO",
    "NO_ORGAO",
    "CPF",
    "nrep_cpf",
    "nvinculos_cpf_orgao"
).agg(
    spark_sum("n_dependentes").alias("total_dependentes"),
    spark_sum(col("n_dependentes") * col("dependente_menor_18")).alias("total_dependentes_menor_18")
).orderBy(
    col("SG_ORGAO"),
    col("nvinculos_cpf_orgao").desc(),
    col('CPF')
)



display(df_vinculo)
```
